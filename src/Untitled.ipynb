```python
from openai import OpenAI
from pydantic import BaseModel
from typing import Optional
import json

client = OpenAI()

class MissionOutput(BaseModel):
    response_status: str
    target_type: str
    module_id: str
    target_resistance: str

    selected_task_id: Optional[str] = None
    mission_title: Optional[str] = None
    target_work: Optional[str] = None
    mission_action: Optional[str] = None
    execution_timing: Optional[str] = None
    execution_method: Optional[str] = None
    ai_tool: Optional[str] = None
    completion_criteria: Optional[str] = None
    mission_rationale: Optional[str] = None
    verification_method: Optional[str] = None
    feedback_rule: Optional[str] = None
    next_action: Optional[str] = None
    no_mission_reason: Optional[str] = None


system_prompt = """
너는 기업 구성원의 AI 활용 행동을 변화시키는 AI Intervention Agent다.

반드시 아래 규칙을 지킨다.

1. Primary Resistance를 변경하지 않는다.
2. Module을 변경하지 않는다.
3. Module의 목적, 목표 행동, 허용 범위, 금지 범위를 준수한다.
4. 별도의 교육이나 업무 외 과제를 생성하지 않는다.
5. 사용자가 실제 수행 중인 Task 안에서만 개입한다.
6. 한 번에 하나의 작은 행동 변화를 우선한다.
7. 실제 사용 가능한 AI Tool만 사용한다.
8. Mission은 무엇을, 언제, 어떻게 수행할지 구체적이어야 한다.
9. 수행 여부를 확인할 수 있는 완료 기준을 제시한다.
10. 적절한 Intervention이 없으면 억지로 Mission을 만들지 않는다.
"""

response = client.responses.parse(
    model="gpt-5.6-terra",
    input=[
        {
            "role": "system",
            "content": system_prompt
        },
        {
            "role": "user",
            "content": json.dumps(
                agent_input,
                ensure_ascii=False
            )
        }
    ],
    text_format=MissionOutput
)

mission = response.output_parsed

print(mission.model_dump_json(indent=2))
```

```python
import pandas as pd
import json

dummy_file = "아모레퍼시픽_Copilot_전사더미데이터_raw_v4(직군세분화).xlsx"
module_file = "AI_Agent_Module_Library_완성본_1.xlsx"

# 테스트 팀
team_id = "T001"

# ----------------------------
# TEAM MASTER
# ----------------------------

team_df = pd.read_excel(
    dummy_file,
    sheet_name="팀마스터"
)

team = team_df[
    team_df["팀ID"] == team_id
].iloc[0]

print("선택 팀:", team["팀명"])


# ----------------------------
# TEAM CONTEXT
# ----------------------------

leadership_df = pd.read_excel(
    dummy_file,
    sheet_name="Leadership_Log"
)

workflow_df = pd.read_excel(
    dummy_file,
    sheet_name="Workflow_Meta"
)

collaboration_df = pd.read_excel(
    dummy_file,
    sheet_name="Collaboration_Log"
)

leadership_team = leadership_df[
    leadership_df["팀ID"] == team_id
]

workflow_team = workflow_df[
    workflow_df["팀ID"] == team_id
]

collaboration_team = collaboration_df[
    collaboration_df["팀ID"] == team_id
]


# ----------------------------
# M13 MODULE
# ----------------------------

module_df = pd.read_excel(
    module_file,
    sheet_name="M13 Leader Role Modeling",
    header=1
)

module_settings = dict(
    zip(
        module_df["Attribute"],
        module_df["설정값"]
    )
)

print("Module:", module_settings.get("Module Name"))
```

```python
from openai import OpenAI
from pydantic import BaseModel
from typing import Optional

client = OpenAI()


class TeamInterventionOutput(BaseModel):

    response_status: str

    module_id: str
    target_resistance: str

    intervention_title: str

    # 팀 전체 알림
    team_notification: str

    # 실제 개입 대상
    intervention_target: str

    # 대상에게 전달할 구체적 행동
    target_action: str

    execution_timing: str

    completion_criteria: str

    verification_method: str

    leader_dashboard_summary: str

    expected_change: str

    no_intervention_reason: Optional[str] = None


team_system_prompt = """
너는 조직의 AI 활용 행동을 변화시키는 Team AI Intervention Agent다.

Team Intervention의 목적은 특정 개인의 AI 사용량 증가가 아니라
팀의 업무방식 자체를 변화시키는 것이다.

반드시 다음 원칙을 지킨다.

1. Primary Team Resistance를 변경하지 않는다.
2. 지정된 Module을 변경하지 않는다.
3. Module의 목적·목표행동·허용범위·금지범위를 준수한다.
4. 별도의 교육이나 새로운 회의를 무조건 추가하지 않는다.
5. 가능한 경우 기존 회의·보고·협업 Workflow 안에서 개입한다.
6. 개입 대상은 리더·Power User·특정 구성원·팀 전체 중
   Resistance 원인에 가장 적합한 대상을 선정한다.
7. 팀 전체에게는 개입의 목적과 역할을 짧게 안내한다.
8. 실제 개입 대상에게는 구체적인 행동 Mission을 제공한다.
9. Mission은 무엇을 언제 어떻게 해야 하는지 명확해야 한다.
10. 팀장이 진행상황을 확인할 수 있도록 관리용 요약도 생성한다.
"""


team_response = client.responses.parse(
    model="gpt-5.6-terra",
    input=[
        {
            "role": "system",
            "content": team_system_prompt
        },
        {
            "role": "user",
            "content": json.dumps(
                team_agent_input,
                ensure_ascii=False,
                default=str
            )
        }
    ],
    text_format=TeamInterventionOutput
)

team_mission = team_response.output_parsed

print(
    team_mission.model_dump_json(
        indent=2
    )
)
```

```python
import pandas as pd
import numpy as np

dummy_file = "아모레퍼시픽_Copilot_전사더미데이터_raw_v4(직군세분화).xlsx"

team_id = "T001"


# ============================================================
# 1. DATA LOAD
# ============================================================

employee_df = pd.read_excel(dummy_file, sheet_name="직원마스터")
pulse_df = pd.read_excel(dummy_file, sheet_name="Pulse_Check")
leadership_df = pd.read_excel(dummy_file, sheet_name="Leadership_Log")
collaboration_df = pd.read_excel(dummy_file, sheet_name="Collaboration_Log")
workflow_df = pd.read_excel(dummy_file, sheet_name="Workflow_Meta")
adoption_df = pd.read_excel(dummy_file, sheet_name="Adoption_Log")
module_map_df = pd.read_excel(dummy_file, sheet_name="처방_모듈맵")
team_df = pd.read_excel(dummy_file, sheet_name="팀마스터")


# ============================================================
# 2. HELPER FUNCTIONS
# ============================================================

# Likert 1~5 → Resistance 0~100
def direct_likert(value):
    """
    정방향:
    1=0, 2=25, 3=50, 4=75, 5=100
    """
    return (value - 1) / 4 * 100


def reverse_likert(value):
    """
    역방향:
    1=100, 2=75, 3=50, 4=25, 5=0
    """
    return (5 - value) / 4 * 100


def positive_metric_to_resistance(value, goal):
    """
    높을수록 좋은 행동지표를 Resistance로 변환.
    goal 이상이면 Resistance = 0
    0이면 Resistance = 100

    ※ goal은 PoC 가정값
    """
    if pd.isna(value):
        return np.nan

    score = 100 * (1 - float(value) / goal)

    return max(
        0,
        min(100, score)
    )


def binary_positive_to_resistance(value):
    """
    긍정 상태가 있으면 Resistance 0,
    없거나 낮음이면 100
    """
    positive_values = [
        "있음",
        "포함",
        "Y",
        "Yes",
        "TRUE",
        True
    ]

    return 0 if value in positive_values else 100


# ============================================================
# 3. TEAM MEMBERS
# ============================================================

team_members = employee_df[
    employee_df["팀ID"] == team_id
][["직원ID"]].copy()

team_member_ids = team_members["직원ID"].tolist()


# ============================================================
# 4. TEAM PULSE DATA
# ============================================================

team_pulse = pulse_df[
    (pulse_df["직원ID"].isin(team_member_ids)) &
    (pulse_df["응답여부"] == "Y")
].copy()


def team_pulse_pair(q_direct, q_reverse):
    """
    팀 구성원의 2개 Pulse 문항을 Resistance Score로 변환 후 평균.

    첫 문항: 정방향
    두 번째 문항: 역방향
    """

    direct_scores = direct_likert(
        pd.to_numeric(
            team_pulse[q_direct],
            errors="coerce"
        )
    )

    reverse_scores = reverse_likert(
        pd.to_numeric(
            team_pulse[q_reverse],
            errors="coerce"
        )
    )

    individual_pair = (
        direct_scores + reverse_scores
    ) / 2

    return individual_pair.mean()


# ============================================================
# 5. TEAM LOG DATA
# ============================================================

lead = leadership_df[
    leadership_df["팀ID"] == team_id
].iloc[0]

collab = collaboration_df[
    collaboration_df["팀ID"] == team_id
].iloc[0]

wf_team = workflow_df[
    workflow_df["팀ID"] == team_id
].copy()

team_adoption = adoption_df[
    adoption_df["팀ID"] == team_id
].copy()


# ============================================================
# 6. SUPPORTING METRICS
# ============================================================

# 팀 전체 AI 활용업무비율 평균
team_ai_usage_rate = (
    team_adoption["AI활용업무비율(%)"]
    .mean()
)

# 표준 Workflow 중 AI 포함 비율
workflow_ai_standard_rate = (
    wf_team["표준프로세스AI단계"]
    .eq("포함")
    .mean()
    * 100
)


# ============================================================
# 7. TEAM RESISTANCE SCORES
#    40% Pulse + 30% Behavior + 30% Behavior
# ============================================================

team_scores = []


def add_score(
    factor,
    gap,
    pulse_score,
    indicator2,
    indicator3,
    indicator2_name,
    indicator3_name
):

    final_score = (
        pulse_score * 0.40
        + indicator2 * 0.30
        + indicator3 * 0.30
    )

    team_scores.append({
        "Gap": gap,
        "Resistance Factor": factor,

        "Pulse Resistance": round(
            pulse_score,
            1
        ),

        indicator2_name: round(
            indicator2,
            1
        ),

        indicator3_name: round(
            indicator3,
            1
        ),

        "Resistance Score": round(
            final_score,
            1
        )
    })


# ------------------------------------------------------------
# ① Role Modeling Gap
# ------------------------------------------------------------

add_score(

    "Role Modeling Gap",
    "Leadership Gap",

    team_pulse_pair(
        "Q25",
        "Q26"
    ),

    positive_metric_to_resistance(
        lead["팀장AI활용업무비율(%)"],
        60
    ),

    positive_metric_to_resistance(
        lead["AI기반회의운영비율(%)"],
        50
    ),

    "Leader AI Usage Resistance",
    "AI Meeting Resistance"
)


# ------------------------------------------------------------
# ② Direction Gap
# ------------------------------------------------------------

add_score(

    "Direction Gap",
    "Leadership Gap",

    team_pulse_pair(
        "Q27",
        "Q28"
    ),

    positive_metric_to_resistance(
        lead["AI방향제시·우선순위(회/월)"],
        6
    ),

    positive_metric_to_resistance(
        lead["AI기반회의운영비율(%)"],
        50
    ),

    "Direction Frequency Resistance",
    "AI Meeting Resistance"
)


# ------------------------------------------------------------
# ③ Reinforcement Gap
# ------------------------------------------------------------

add_score(

    "Reinforcement Gap",
    "Leadership Gap",

    team_pulse_pair(
        "Q29",
        "Q30"
    ),

    positive_metric_to_resistance(
        lead["AI활용독려·피드백(회/월)"],
        6
    ),

    positive_metric_to_resistance(
        lead["AI활용사례공유·인정(회/월)"],
        5
    ),

    "Feedback Resistance",
    "Recognition Resistance"
)


# ------------------------------------------------------------
# ④ Psychological Safety Gap
# ------------------------------------------------------------

add_score(

    "Psychological Safety Gap",
    "Leadership Gap",

    team_pulse_pair(
        "Q31",
        "Q32"
    ),

    binary_positive_to_resistance(
        lead["실수·질문 허용 분위기"]
    ),

    # PoC Proxy:
    # 정기적인 피드백이 Safety 형성에 기여한다고 가정
    positive_metric_to_resistance(
        lead["AI활용독려·피드백(회/월)"],
        6
    ),

    "Safety Climate Resistance",
    "Support Feedback Resistance"
)


# ------------------------------------------------------------
# ⑤ Sharing Gap
# ------------------------------------------------------------

add_score(

    "Sharing Gap",
    "Collaboration Gap",

    team_pulse_pair(
        "Q33",
        "Q34"
    ),

    positive_metric_to_resistance(
        collab["AI경험공유(회/월)"],
        6
    ),

    positive_metric_to_resistance(
        collab["노하우자산등록(건/월)"],
        5
    ),

    "Experience Sharing Resistance",
    "Knowledge Registration Resistance"
)


# ------------------------------------------------------------
# ⑥ Knowledge Diffusion Gap
# ------------------------------------------------------------

add_score(

    "Knowledge Diffusion Gap",
    "Collaboration Gap",

    team_pulse_pair(
        "Q35",
        "Q36"
    ),

    positive_metric_to_resistance(
        collab["노하우자산등록(건/월)"],
        5
    ),

    positive_metric_to_resistance(
        collab["자산재사용(건/월)"],
        6
    ),

    "Knowledge Registration Resistance",
    "Knowledge Reuse Resistance"
)


# ------------------------------------------------------------
# ⑦ Peer Learning Gap
# ------------------------------------------------------------

add_score(

    "Peer Learning Gap",
    "Collaboration Gap",

    team_pulse_pair(
        "Q37",
        "Q38"
    ),

    positive_metric_to_resistance(
        collab["동료페어학습(회/월)"],
        4
    ),

    positive_metric_to_resistance(
        collab["AI경험공유(회/월)"],
        6
    ),

    "Peer Learning Resistance",
    "Experience Sharing Resistance"
)


# ------------------------------------------------------------
# ⑧ Collective Adoption Gap
# ------------------------------------------------------------

add_score(

    "Collective Adoption Gap",
    "Collaboration Gap",

    team_pulse_pair(
        "Q39",
        "Q40"
    ),

    positive_metric_to_resistance(
        team_ai_usage_rate,
        60
    ),

    binary_positive_to_resistance(
        collab["공동Workflow표준화"]
    ),

    "Team AI Usage Resistance",
    "Workflow Standardization Resistance"
)


# ============================================================
# 8. RESULT TABLE
# ============================================================

team_resistance_df = pd.DataFrame(
    team_scores
)

team_resistance_df = (
    team_resistance_df
    .sort_values(
        "Resistance Score",
        ascending=False
    )
    .reset_index(drop=True)
)


# ============================================================
# 9. PRIMARY TEAM RESISTANCE
# ============================================================

primary_row = team_resistance_df.iloc[0]

primary_team_resistance = (
    primary_row["Resistance Factor"]
)

primary_team_score = (
    primary_row["Resistance Score"]
)


# ============================================================
# 10. MODULE AUTO MATCHING
# ============================================================

module_row = module_map_df[
    module_map_df["세부Factor"]
    == primary_team_resistance
].iloc[0]

assigned_module = (
    module_row["처방Module"]
)

core_approach = (
    module_row["핵심접근법"]
)


# ============================================================
# 11. SCORE LEVEL
# ============================================================

def resistance_level(score):

    if score >= 70:
        return "High"

    elif score >= 40:
        return "Medium"

    return "Low"


primary_level = resistance_level(
    primary_team_score
)


# ============================================================
# 12. OUTPUT
# ============================================================

team_name = team_df[
    team_df["팀ID"] == team_id
]["팀명"].iloc[0]


print("\n==========================================")
print("TEAM RESISTANCE DIAGNOSIS")
print("==========================================")

print("팀:", team_name)

print(
    f"팀 Copilot 활용률: "
    f"{team_ai_usage_rate:.1f}%"
)

print()

print(
    team_resistance_df[
        [
            "Gap",
            "Resistance Factor",
            "Pulse Resistance",
            "Resistance Score"
        ]
    ].to_string(
        index=False
    )
)

print("\n==========================================")
print("PRIMARY TEAM RESISTANCE")
print("==========================================")

print(
    "Primary Resistance:",
    primary_team_resistance
)

print(
    "Resistance Score:",
    primary_team_score
)

print(
    "Level:",
    primary_level
)

print(
    "Assigned Module:",
    assigned_module
)

print(
    "Core Approach:",
    core_approach
)
```

```python
from openai import OpenAI
from pydantic import BaseModel
from typing import Optional

client = OpenAI()


class TeamMissionOutput(BaseModel):

    response_status: str

    module_id: str
    target_resistance: str

    # 팀 전체에게 보여줄 내용
    team_summary_title: str
    improvement_part: str
    team_action: str

    # 실제 개입 대상
    intervention_target: str
    target_action: str

    execution_timing: str
    completion_criteria: str
    verification_method: str

    expected_change: str

    no_mission_reason: Optional[str] = None


system_prompt = """
너는 기업 조직의 AI 활용 행동을 변화시키는
Team AI Intervention Agent다.

반드시 아래 규칙을 따른다.

1. Primary Team Resistance를 변경하지 않는다.
2. Module을 변경하지 않는다.
3. Module의 목적과 목표 행동을 변경하지 않는다.
4. Module Library의 허용 Intervention 안에서만 개입한다.
5. 금지 Intervention은 절대 생성하지 않는다.
6. 별도의 교육, 발표, 과제를 임의로 추가하지 않는다.
7. 기존 팀 업무 Workflow 안에 개입을 삽입한다.
8. 팀의 AI 활용률과 실제 협업 행동 데이터를 고려한다.
9. Team Intervention의 목적은 개인의 사용량이 아니라
   팀 업무방식 자체의 변화를 만드는 것이다.
10. 팀 전체 안내문은 매우 짧게 생성한다.
11. 실제 개입 대상에게는 구체적 행동을 별도로 생성한다.
12. 무엇을, 언제, 어떻게 수행할지가 명확해야 한다.
13. 수행 여부를 확인할 수 있는 완료 기준을 만든다.
14. 적절한 개입 기회가 없다면 Mission을 억지로 생성하지 않는다.
"""

response = client.responses.parse(
    model="gpt-5.6-terra",
    input=[
        {
            "role": "system",
            "content": system_prompt
        },
        {
            "role": "user",
            "content": json.dumps(
                team_context,
                ensure_ascii=False
            )
        }
    ],
    text_format=TeamMissionOutput
)

team_mission = response.output_parsed

print(
    team_mission.model_dump_json(
        indent=2
    )
)
```

```python
import tkinter as tk
from tkinter import font


# ============================================================
# TEAM INTERVENTION POPUP
# GPT team_mission 결과 자동 연결
# ============================================================

def show_team_intervention():

    root = tk.Tk()

    root.title("AI Work Assistant")

    # 팝업 크기
    width = 620
    height = 470

    # 화면 오른쪽 아래 배치
    screen_width = root.winfo_screenwidth()
    screen_height = root.winfo_screenheight()

    x = screen_width - width - 40
    y = screen_height - height - 90

    root.geometry(
        f"{width}x{height}+{x}+{y}"
    )

    root.configure(bg="#F4F7F9")

    root.resizable(False, False)

    # 항상 위
    root.attributes("-topmost", True)


    # ========================================================
    # MAIN CARD
    # ========================================================

    card = tk.Frame(
        root,
        bg="white",
        padx=30,
        pady=25
    )

    card.pack(
        fill="both",
        expand=True,
        padx=16,
        pady=16
    )


    # ========================================================
    # HEADER
    # ========================================================

    header = tk.Frame(
        card,
        bg="white"
    )

    header.pack(
        fill="x"
    )


    tk.Label(
        header,
        text="✦  TEAM AI INTERVENTION",
        font=("맑은 고딕", 11, "bold"),
        fg="#159CA5",
        bg="white"
    ).pack(
        side="left"
    )


    tk.Label(
        header,
        text="지금",
        font=("맑은 고딕", 9),
        fg="#9AA6AF",
        bg="white"
    ).pack(
        side="right"
    )


    # ========================================================
    # TEAM NAME
    # ========================================================

    tk.Label(
        card,
        text=team_name,
        font=("맑은 고딕", 10),
        fg="#84919B",
        bg="white"
    ).pack(
        anchor="w",
        pady=(22, 4)
    )


    # ========================================================
    # 1. COPILOT USAGE
    # ========================================================

    usage_frame = tk.Frame(
        card,
        bg="#F4FAFA",
        padx=18,
        pady=14
    )

    usage_frame.pack(
        fill="x",
        pady=(6, 8)
    )


    tk.Label(
        usage_frame,
        text="팀 Copilot 활용률",
        font=("맑은 고딕", 9),
        fg="#687985",
        bg="#F4FAFA"
    ).pack(
        side="left"
    )


    tk.Label(
        usage_frame,
        text=f"{team_ai_usage_rate:.1f}%",
        font=("맑은 고딕", 18, "bold"),
        fg="#159CA5",
        bg="#F4FAFA"
    ).pack(
        side="right"
    )


    # ========================================================
    # 2. IMPROVEMENT PART
    # ========================================================

    improvement_frame = tk.Frame(
        card,
        bg="#FFF8F0",
        padx=18,
        pady=13
    )

    improvement_frame.pack(
        fill="x",
        pady=8
    )


    tk.Label(
        improvement_frame,
        text="개선 포인트",
        font=("맑은 고딕", 9),
        fg="#8A7562",
        bg="#FFF8F0"
    ).pack(
        anchor="w"
    )


    tk.Label(
        improvement_frame,
        text=team_mission.improvement_part,
        font=("맑은 고딕", 12, "bold"),
        fg="#18364C",
        bg="#FFF8F0",
        wraplength=510,
        justify="left"
    ).pack(
        anchor="w",
        pady=(4, 0)
    )


    # ========================================================
    # 3. TEAM ACTION
    # ========================================================

    action_frame = tk.Frame(
        card,
        bg="#F5F7FA",
        padx=18,
        pady=13
    )

    action_frame.pack(
        fill="x",
        pady=8
    )


    tk.Label(
        action_frame,
        text="이번에 해야 할 일",
        font=("맑은 고딕", 9),
        fg="#687985",
        bg="#F5F7FA"
    ).pack(
        anchor="w"
    )


    tk.Label(
        action_frame,
        text=team_mission.team_action,
        font=("맑은 고딕", 11, "bold"),
        fg="#18364C",
        bg="#F5F7FA",
        wraplength=510,
        justify="left"
    ).pack(
        anchor="w",
        pady=(4, 0)
    )


    # ========================================================
    # BUTTON ACTIONS
    # ========================================================

    def later():

        print(
            f"[TEAM INTERVENTION] "
            f"{team_name} → LATER"
        )

        root.destroy()


    def confirm():

        print(
            f"[TEAM INTERVENTION] "
            f"{team_name} → CONFIRMED"
        )

        print(
            "Target:",
            team_mission.intervention_target
        )

        print(
            "Action:",
            team_mission.target_action
        )

        root.destroy()


    # ========================================================
    # BUTTONS
    # ========================================================

    button_frame = tk.Frame(
        card,
        bg="white"
    )

    button_frame.pack(
        fill="x",
        pady=(20, 0)
    )


    later_button = tk.Button(
        button_frame,
        text="나중에",
        command=later,

        bg="#EEF2F4",
        fg="#526573",

        activebackground="#E3E8EB",

        relief="flat",
        bd=0,

        font=("맑은 고딕", 10, "bold"),

        height=2,

        cursor="hand2"
    )

    later_button.pack(
        side="left",
        fill="x",
        expand=True,
        padx=(0, 7)
    )


    confirm_button = tk.Button(
        button_frame,
        text="확인했어요",
        command=confirm,

        bg="#153D58",
        fg="white",

        activebackground="#23536F",
        activeforeground="white",

        relief="flat",
        bd=0,

        font=("맑은 고딕", 10, "bold"),

        height=2,

        cursor="hand2"
    )

    confirm_button.pack(
        side="left",
        fill="x",
        expand=True,
        padx=(7, 0)
    )


    root.mainloop()


# ============================================================
# 실행
# ============================================================

show_team_intervention()
```
